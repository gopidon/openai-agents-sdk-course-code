## Cell 1 — Python Version Check

Google Colab ships with a recent Python version, but let's confirm it meets the minimum requirement.

The OpenAI Agents SDK requires **Python 3.9 or higher**. Run this cell first — if the check fails, switch your Colab runtime before continuing.

```python
import sys
print(sys.version)

major, minor = sys.version_info.major, sys.version_info.minor
if major < 3 or (major == 3 and minor < 9):
    print("\n⚠️  WARNING: Python 3.9 or higher is required.")
    print("   Go to Runtime → Change runtime type and select a newer Python version.")
else:
    print(f"\n✅ Python {major}.{minor} — you're good to go!")
```

## Cell 4 — Set Up the OpenAI API Key

We use **Google Colab Secrets** to load the API key. This is the right approach for Colab: the key is stored securely in your Google account and never written into the notebook file itself, so you can share or download this notebook without exposing your key.

### How to add your key to Colab Secrets

Follow these steps before running the code cell below:

1. Look at the **left sidebar** in Colab. You will see a column of icons. Click the **🔑 key icon** — this opens the Secrets panel. (Alternatively, go to **Tools → Secrets** in the top menu.)
2. In the Secrets panel, click **+ Add new secret**.
3. In the **Name** field, type exactly: `OPENAI_API_KEY`
4. In the **Value** field, paste your OpenAI API key (it starts with `sk-`).
5. Make sure the **Notebook access** toggle is switched **ON** — this grants this specific notebook permission to read the secret.
6. Click **Save** (or press Enter).

Once the secret is saved, run the cell below to load it into the environment.

```python
import os
from google.colab import userdata

os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
print("✅ API key loaded from Colab Secrets.")
```

## Cell 5 — Verify the API Key Is Loaded

This cell confirms the key made it into the environment. It prints only the first eight characters followed by dots — enough to recognise your key without exposing it.

If you see `NOT SET`, go back to Cell 4 and make sure the secret name is spelled exactly `OPENAI_API_KEY` and that Notebook access is toggled ON.

```python
import os

api_key = os.environ.get("OPENAI_API_KEY")

print("API key set:", api_key is not None)
print("Key preview:", api_key[:8] + "..." if api_key else "NOT SET")

if not api_key:
    print("\n⚠️  No API key found. Please complete the Colab Secrets setup in Cell 4 and re-run.")
else:
    print("\n✅ API key is loaded. Ready to connect.")
```

## Cell 6 — Quick Connectivity Test

This is a minimal API call — not an agent run — just to confirm your key actually works and can reach the OpenAI API. If this returns `CONNECTED`, your environment is fully set up and ready for Lecture 1.6.

```python
from openai import AsyncOpenAI

client = AsyncOpenAI()

async def connectivity_test():
    response = await client.chat.completions.create(
        model="gpt-5.4-mini",
        messages=[{"role": "user", "content": "Reply with exactly: CONNECTED"}]
    )
    return response.choices[0].message.content

result = await connectivity_test()
print("Response:", result)
print("\n✅ Connectivity confirmed — your environment is fully set up!")
```
